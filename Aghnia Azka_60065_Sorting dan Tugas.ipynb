Bubble sort

```python
def bubble_sort(arr):
  n = len(arr)
  for i in range(n - 1): # Corrected 'l' to '1'
    for j in range(n - 1 - i): # Corrected 'l' to '1', and adjusted the range with 'i'
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr

# contoh penggunaan
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = bubble_sort(data.copy()) # gunakan copy agar data asli tidak berubah
print("Data sebelum sorting:", data)
print("Data setelah bubble sort:", sorted_data)
```

Selection sort

```python
def selection_sort(arr):
  n = len(arr)
  for i in range(n - 1):
    min_index = i
    for j in range(i + 1, n): # Corrected 'l' to '1'
      if arr[j] < arr[min_index]:
        min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]
  return arr

# contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = selection_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data sesuddah selection sort:", sorted_data)
```

Insertion sort

```python
def insertion_sort(arr):
  n = len(arr)
  for i in range(1, n): # Corrected 'l' to '1'
    key = arr[i]
    j = i - 1 # Corrected 'l' to '1'
    while j >= 0 and arr[j] > key:
      arr[j + 1] = arr[j] # Corrected 'l' to '1'
      j -= 1 # Corrected 'l' to '1'
    arr[j + 1] = key # Corrected 'l' to '1'
  return arr

# contoh penggunaan
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = insertion_sort(data.copy())
print("Data sebelum sorting:", data) # Corrected 'rint' to 'print'
print("Data sesuddah insertion sort:", sorted_data)
```

Merge sort

```python
def merge_sort(arr):
  if len(arr) > 1:
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    merge_sort(left_half)
    merge_sort(right_half)

    i = j = k = 0

    while i < len(left_half) and j < len(right_half):
      if left_half[i] < right_half[j]: # Corrected 'left_halh' to 'left_half'
        arr[k] = left_half[i]
        i += 1 # Corrected 'l' to '1'
      else:
        arr[k] = right_half[j]
        j += 1 # Corrected 'l' to '1'
      k += 1 # Corrected 'l' to '1'

    while i < len(left_half):
      arr[k] = left_half[i]
      i += 1 # Corrected 'l' to '1'
      k += 1 # Corrected 'l' to '1'

    while j < len(right_half):
      arr[k] = right_half[j]
      j += 1 # Corrected 'l' to '1'
      k += 1 # Corrected 'l' to '1'

  return arr

# contoh penggunaan
data = [64, 25, 12, 22, 11] # Corrected 'data + ...' to 'data = ...'
sorted_data = merge_sort(data.copy())
print("Data sebelum sorting:", data) # Corrected 'rint' to 'print'
print("Data sesuddah merge sort:", sorted_data)
```

Quick sort

```python
def quick_sort(arr):
  if len(arr) <= 1:
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)

# contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = quick_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data sesudah quick sort:", sorted_data)
```
