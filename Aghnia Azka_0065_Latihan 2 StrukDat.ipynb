```python
# ==============================
# DOUBLE LINKED LIST
# 3 node + sisip tengah + hapus kata kunci
# ==============================

class Node:
    def __init__(self, dataval):
        self.dataval = dataval
        self.next = None
        self.prev = None


class DoubleLinkedList:
    def __init__(self):
        self.head = None

    def printlist(self):
        temp = self.head
        while temp:
            print(temp.dataval)
            temp = temp.next

    def add_end(self, data):
        newnode = Node(data)
        if self.head is None:
            self.head = newnode
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = newnode
        newnode.prev = temp

    def add_middle(self, prev_node, data):
        if prev_node is None:
            return

        newnode = Node(data)
        newnode.next = prev_node.next
        prev_node.next = newnode
        newnode.prev = prev_node

        if newnode.next:
            newnode.next.prev = newnode

    def remove_key(self, key):
        temp = self.head

        while temp:
            if temp.dataval == key:

                if temp.prev:
                    temp.prev.next = temp.next
                else:
                    self.head = temp.next

                if temp.next:
                    temp.next.prev = temp.prev

                temp = None
                return

            temp = temp.next


print("DOUBLE LINKED LIST")
dataList = DoubleLinkedList()

dataList.add_end("Tanghulu")
dataList.add_end("Mochi")
dataList.add_end("Mango")

dataList.add_middle(dataList.head, "Sandwich")
dataList.remove_key("Mochi")

dataList.printlist()
```
