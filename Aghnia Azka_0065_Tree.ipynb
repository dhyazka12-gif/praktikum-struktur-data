Binary Tree Manual

```python
class Node:
  def __init__ (self, data):
    self.data = data
    self.left = None
    self.right = None

# Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

# Fungsi inorder tranversal
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)

print("Hasil inorder tranversal dari binary tree:")
inorder(root)
```

Binary Search Tree

```python
class Node:
  def __init__ (self, data):
    self.data = data
    self.left = None
    self.right = None

class BinarySearchTree:
  def __init__ (self):
    self.root = None

  def insert(self, root, data):
    if root is None:
      return Node(data)
    if data < root.data:
      root.left = self.insert(root.left, data)
    else:
      root.right = self.insert(root.right, data)
    return root

bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]
for value in data_list:
  root = bst.insert(root, value)
```

Traversal pada BST

```python
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)

def preorder(node):
  if node:
    print(node.data, end=' ')
    preorder(node.left)
    preorder(node.right)

def postorder(node):
  if node:
    postorder(node.left)
    postorder(node.right)
    print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

Pencarian Nilai dalam BST

```python
def search(node, key):
  if node is None or node.data == key:
    return node
  if key < node.data:
    return search(node.left, key)
  return search(node.right, key)

# Uji pencarian
key = 60
result = search(root, key)
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan dalam tree.")
```

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

# Fungsi hitung vokal
def hitung_vokal(nama):
  vokal = "aiueoAIUEO"
  return sum(1 for huruf in nama if huruf in vokal)

def inorder(root):
  if root:
    inorder(root.left)
    print(root.data, end=' ')
    inorder(root.right)

# Ubah data
nama = "Azka Dhiya"
npm = "2505060065"
tanggal_lahir = "12-05"

# Olah data
jumlah_huruf = len(nama.replace(" ", ""))
jumlah_vokal = hitung_vokal(nama)
dua_digit_npm = int(npm[-2:])
tanggal = int(tanggal_lahir.split("-")[0])
bulan = int(tanggal_lahir.split("-")[1])

# Tree
root = Node(jumlah_huruf)

root.left = Node(jumlah_vokal)
root.right = Node(dua_digit_npm)

root.left.left = Node(tanggal)
root.right.right = Node(bulan)

print("Jumlah huruf:", jumlah_huruf)
print("Jumlah vokal:", jumlah_vokal)
print("Dua digit terakhir NPM:", dua_digit_npm)
print("Tanggal lahir:", tanggal)
print("Bulan lahir:", bulan)

print("\nInorder Traversal:")
inorder(root)
```

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

# Fungsi insert BST
def insert(root, data):
  if root is None:
    return Node(data)

  # ubah ke int saat dibandingkan
  if int(data) < int(root.data):
    root.left = insert(root.left, data)

  elif int(data) > int(root.data):
    root.right = insert(root.right, data)

  return root

# Traversal
def inorder(root):
  if root:
    inorder(root.left)
    print(root.data, end=" ")
    inorder(root.right)

def preorder(root):
  if root:
    print(root.data, end=" ")
    preorder(root.left)
    preorder(root.right)

def postorder(root):
  if root:
    postorder(root.left)
    postorder(root.right)
    print(root.data, end=" ")

# Fungsi search
def search(root, key):
  if root is None:
    return False

  if root.data == key:
    return True

  elif int(key) < int(root.data):
    return search(root.left, key)

  else:
    return search(root.right, key)

npm = "2505060065"
nama = "Azka Dhiya"

tiga_digit_npm = npm[-3:]   # tetap string → "065"
dua_digit_npm = npm[-2:]    # "65"

tanggal = "12"
bulan = "05"

ascii1 = str(ord(nama[0]))
ascii2 = str(ord(nama[1]))

# dua angka bebas
fav1 = "18"
fav2 = "80"

# List data
data = [
    tiga_digit_npm,
    tanggal,
    bulan,
    ascii1,
    ascii2,
    fav1,
    fav2
]

print("3 Digit NPM   :", tiga_digit_npm)
print("Tanggal       :", tanggal)
print("Bulan         :", bulan)
print("ASCII 1       :", ascii1)
print("ASCII 2       :", ascii2)
print("Fav 1         :", fav1)
print("Fav 2         :", fav2)

root = None

for angka in data:
  root = insert(root, angka)

print("\nInorder   :", end=" ")
inorder(root)

print("\nPreorder  :", end=" ")
preorder(root)

print("\nPostorder :", end=" ")
postorder(root)

print("\n\nPencarian:")

if search(root, dua_digit_npm):
  print(dua_digit_npm, "ditemukan.")
else:
  print(dua_digit_npm, "tidak ditemukan.")

cari_lain = "99"

if search(root, cari_lain):
  print(cari_lain, "ditemukan.")
else:
  print(cari_lain, "tidak ditemukan.")
```
